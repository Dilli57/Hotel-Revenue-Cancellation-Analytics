# Monthly Booking Trend

```python
plt.figure(figsize=(12,5))

ax = sns.countplot(x='arrival_date_month', data=df)

# Add labels on top of bars
for p in ax.patches:
    ax.annotate(f'{int(p.get_height())}', 
                (p.get_x() + p.get_width() / 2., p.get_height()),
                ha='center', va='bottom',
                fontsize=10)

plt.title("Monthly Bookings")
plt.xticks(rotation=45)

plt.show()
```

# ADR (Average Daily Rate) Analysis

```python
plt.figure(figsize=(10,5))

sns.histplot(df[df['adr'] < 500]['adr'], bins=50)

plt.title("ADR Distribution (Without Outliers)")

plt.xlabel("ADR")
plt.ylabel("Count")

plt.show()
```

# Market Segment Analysis

```python
plt.figure(figsize=(12,5))

ax = sns.countplot(x='market_segment', data=df)

# Add labels on bars
for p in ax.patches:
    ax.annotate(f'{int(p.get_height())}',
                (p.get_x() + p.get_width()/2, p.get_height()),
                ha='center',
                va='bottom',
                fontsize=9)

plt.xticks(rotation=45)

plt.title("Market Segment Distribution")
plt.xlabel("Market Segment")
plt.ylabel("Count")

plt.show()
```

# Feature Engineering

# Total Guests

```python
df['total_guests'] = df['adults'] + df['children'] + df['babies']
```

# Total Nights Stayed

```python
df['total_nights'] = (
    df['stays_in_weekend_nights'] +
    df['stays_in_week_nights']
)
```

# Total Revenue

```python
df['total_revenue'] = df['adr'] * df['total_nights']
```

# Booking Month Number

```python
df['reservation_month'] = df['reservation_status_date'].dt.month
```
